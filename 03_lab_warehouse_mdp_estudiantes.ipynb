## Reglas del mundo

El grid tiene **5 filas × 6 columnas**.

### Estados especiales

A partir de la imagen identifica:

- estación de carga `+2` (**terminal**);

### Recompensa

Usamos la convención del notebook de clase, es decir, **\(R(s)\)**:

- entrega: `+10`;
- carga: `+2`;
- peligro mortal: `-10`;
- peligro: `-3`;
- cualquier otro estado transitable: `-1` (costo por paso).

### Dinámica

La transición depende del **estado actual**:

$$

$$

$$

$$

## Parte 1 — Modela el MDP

Completa la clase `WarehouseMDP`.

La parte importante no es escribir muchas líneas de código: es traducir correctamente la imagen a:

- estados;
- acciones;
- recompensas;
- terminales;
- obstáculos;
- tipos de piso;
- función de transición.


````python
import numpy as np

class WarehouseMDP:
    def __init__(self):
        self.height = 5
        self.width = 6

        # --- Layout leído de la imagen ---
        self.start = (0, 0)
        self.walls = {(0, 3), (1, 1), (2, 4), (4, 2)}      # estanterías
        self.slippery_states = {(1, 2), (2, 1), (3, 3)}    # piso resbaloso

        self.terminal_states = {
            (0, 5):  10.0,   # entrega
            (2, 2):   2.0,   # carga
            (3, 5): -10.0,   # peligro mortal
        }
        self.danger_states = {
            (1, 4): -3.0,
            (4, 1): -3.0,
        }

        self.living_reward = -1.0
        self.gamma = 0.9

        self.actions = [
            (-1, 0),  # UP
            ( 1, 0),  # DOWN
            ( 0,-1),  # LEFT
            ( 0, 1),  # RIGHT
        ]

    def is_valid_state(self, state):
        # Revisa que no se halla quedado en una pared
        r, c = state
        return (0 <= r < self.height and 0 <= c < self.width
                and state not in self.walls)

    def states(self):
        # Todas las celdas dentro del grid que NO sean pared.
        # Las terminales SÍ son estados (posiciones válidas y absorbentes).
        return [(r, c) for r in range(self.height) for c in range(self.width)
                if (r, c) not in self.walls]

    def is_terminal(self, state):
        return state in self.terminal_states

    def get_reward(self, state):
        # R(s): prioridad terminal -> peligro -> costo por paso
        if state in self.terminal_states:
            return self.terminal_states[state]
        if state in self.danger_states:
            return self.danger_states[state]
        return self.living_reward

    def get_transition_probs(self, state, action):
        """
        Devuelve [(next_state, probability), ...].
        - Probabilidades según el piso del estado ACTUAL (resbaloso o no).
        - Borde o pared -> el robot permanece en 'state'.
        - Terminales -> absorbentes (self-loop), para que T(s,a,·) sume 1.
        """
        if self.is_terminal(state):
            return [(state, 1.0)]

        if state in self.slippery_states:
            p_main, p_side = 0.60, 0.20
        else:
            p_main, p_side = 0.90, 0.05

        dr, dc = action
        outcomes = [
            (action,     p_main),   # dirección elegida
            ((dc, -dr),  p_side),   # perpendicular 1  (izq/der: misma prob)
            ((-dc, dr),  p_side),   # perpendicular 2
        ]

        probs = {}
        for (mr, mc), p in outcomes:
            nxt = (state[0] + mr, state[1] + mc)
            if not self.is_valid_state(nxt):   # fuera del grid o estantería
                nxt = state                     # -> se queda en su sitio
            probs[nxt] = probs.get(nxt, 0.0) + p

        return list(probs.items())
````


### Validación mínima del modelo

Antes de implementar Bellman, valida primero el MDP.


````python
grid = WarehouseMDP()

S = grid.states()
print("Número de estados:", len(S))

# Cada distribución T(s,a,·) debe sumar 1.
for s in S:
    for a in grid.actions:
        transitions = grid.get_transition_probs(s, a)
        total = sum(p for _, p in transitions)
        assert abs(total - 1.0) < 1e-12

print("✓ Todas las distribuciones de transición suman 1.")
````


## Parte 2 — Value Iteration

Implementa:

$$
V_{k+1}(s)
=
R(s)+\gamma\max_a
\sum_{s'}T(s,a,s')V_k(s')
$$


````python
def expected_next_value(grid, state, action, V):
    # sum_{s'} T(s,a,s') V(s')
    return sum(p * V[s2] for s2, p in grid.get_transition_probs(state, action))


def value_iteration(grid, threshold=1e-4, max_iter=10_000):
    V = {s: 0.0 for s in grid.states()}
    for s in grid.states():                      # terminales fijas en R(s)
        if grid.is_terminal(s):
            V[s] = grid.get_reward(s)

    for iterations in range(1, max_iter + 1):
        delta = 0.0
        V_new = dict(V)                          # actualización síncrona (V_{k+1} desde V_k)
        for s in grid.states():
            if grid.is_terminal(s):
                continue                         # no se actualiza: V(terminal) = R(terminal)
            best = max(expected_next_value(grid, s, a, V) for a in grid.actions)
            V_new[s] = grid.get_reward(s) + grid.gamma * best
            delta = max(delta, abs(V_new[s] - V[s]))
        V = V_new
        if delta < threshold:
            break

    return V, iterations


def extract_policy(grid, V):
    policy = {}
    for s in grid.states():
        if grid.is_terminal(s):
            continue                             # las terminales no tienen acción
        policy[s] = max(grid.actions,
                        key=lambda a: expected_next_value(grid, s, a, V))
    return policy
````


## Parte 3 — Policy Iteration

### Policy Evaluation

$$
V_{k+1}^{\pi}(s)
=
R(s)+\gamma
\sum_{s'}T(s,\pi(s),s')V_k^\pi(s')
$$

### Policy Improvement

$$
\pi_{\mathrm{new}}(s)
=
\arg\max_a
\sum_{s'}T(s,a,s')V^\pi(s')
$$

````python
def policy_evaluation(grid, policy, threshold=1e-4, max_iter=10_000):
    # Resuelve V^pi para una política FIJA. Diferencia clave con Value Iteration:
    # aquí NO hay max sobre acciones; se usa siempre la acción que dicta 'policy'.
    # Es el sistema lineal de Bellman para pi, resuelto de forma iterativa.
    V = {s: 0.0 for s in grid.states()}
    for s in grid.states():
        if grid.is_terminal(s):
            V[s] = grid.get_reward(s)          # terminales fijas en R(s)

    for _ in range(max_iter):
        delta = 0.0
        V_new = dict(V)
        for s in grid.states():
            if grid.is_terminal(s):
                continue
            # V^pi(s) = R(s) + gamma * sum_{s'} T(s, pi(s), s') V^pi(s')
            V_new[s] = grid.get_reward(s) + grid.gamma * expected_next_value(grid, s, policy[s], V)
            delta = max(delta, abs(V_new[s] - V[s]))
        V = V_new
        if delta < threshold:
            break
    return V


def policy_improvement(grid, V):
    # Dado V^pi, elige en cada estado la acción greedy:
    # pi_new(s) = argmax_a sum_{s'} T(s,a,s') V(s')
    # (Idéntica en forma a extract_policy: la "mejora" es un paso greedy de un solo nivel.)
    policy = {}
    for s in grid.states():
        if grid.is_terminal(s):
            continue                            # las terminales no tienen acción
        policy[s] = max(grid.actions,
                        key=lambda a: expected_next_value(grid, s, a, V))
    return policy


def policy_iteration(grid, threshold=1e-4, max_iter=100):
    # 1. Política inicial arbitraria: UP en todos los estados no terminales.
    policy = {s: grid.actions[0] for s in grid.states() if not grid.is_terminal(s)}
    history = []                                # nº de estados que cambian de acción por iteración

    V = None
    for _ in range(max_iter):
        # 2. Evaluación: resolver V^pi de la política ACTUAL (fija).
        V = policy_evaluation(grid, policy, threshold)
        # 3. Mejora: recalcular la acción greedy en cada estado.
        new_policy = policy_improvement(grid, V)
        # 4. Prueba de estabilidad: si ningún estado cambió de acción, la política es óptima
        #    (teorema de mejora de políticas: iterar eval+mejora converge a pi* en pasos finitos).
        changes = sum(1 for s in new_policy if new_policy[s] != policy[s])
        history.append(changes)
        policy = new_policy
        if changes == 0:
            break

    return policy, V, history
````


## Parte 4 — Visualización y comparación


````python
ARROWS = {
    (-1, 0): "↑",
    ( 1, 0): "↓",
    ( 0,-1): "←",
    ( 0, 1): "→",
}

def print_values(grid, V):
    for r in range(grid.height):
        row = []
        for c in range(grid.width):
            s = (r, c)
            if s in grid.walls:
                row.append("  WALL  ")
            else:
                row.append(f"{V[s]:+7.3f}")
        print(" | ".join(row))


def print_policy(grid, policy):
    for r in range(grid.height):
        row = []
        for c in range(grid.width):
            s = (r, c)

            if s in grid.walls:
                row.append(" # ")
            elif grid.is_terminal(s):
                reward = grid.get_reward(s)
                row.append(f"{reward:+.0f}")
            else:
                row.append(f" {ARROWS[policy[s]]} ")

        print(" | ".join(row))
````

````python
# VALUE ITERATION
V_vi, n_vi = value_iteration(grid)
pi_vi = extract_policy(grid, V_vi)

print("=== VALUE ITERATION ===")
print("Iteraciones:", n_vi)
print("\nValores:")
print_values(grid, V_vi)
print("\nPolítica:")
print_policy(grid, pi_vi)


# POLICY ITERATION
pi_pi, V_pi, history = policy_iteration(grid)

print("\n=== POLICY ITERATION ===")
print("Historia:", history)
print("\nValores:")
print_values(grid, V_pi)
print("\nPolítica:")
print_policy(grid, pi_pi)

assert pi_vi == pi_pi
print("\n✓ Ambos algoritmos encontraron la misma política óptima.")
````
